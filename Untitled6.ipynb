```python
import os
import random
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow import keras
from tensorflow.keras import layers

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("TensorFlow version:", tf.__version__)

print("Available GPU:", tf.config.list_physical_devices('GPU'))

print("Setup completed successfully!")
```

```python
from google.colab import files
import zipfile
import os
import shutil

uploaded = files.upload()

zip_path = list(uploaded.keys())[0]

extract_path = "/content/dataset"

if os.path.exists(extract_path):
    shutil.rmtree(extract_path)

os.makedirs(extract_path)

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall(extract_path)

print("Dataset extracted successfully!")
print("Extracted folders:", os.listdir(extract_path))
```

```python
dataset_path = "/content/dataset/Laptop_Dataset/Laptop_Dataset"

classes = ["Laptops", "Not_Laptops"]
valid_extensions = (".jpg", ".jpeg", ".png", ".webp")

total_images = 0

for class_name in classes:

    class_path = os.path.join(dataset_path, class_name)

    images = [
        file for file in os.listdir(class_path)
        if file.lower().endswith(valid_extensions)
    ]

    count = len(images)
    total_images += count

    print(f"{class_name}: {count} images")

print("-" * 35)
print("Total images:", total_images)
```

```python
import matplotlib.pyplot as plt
import random
from PIL import Image

plt.figure(figsize=(15, 8))

for i, class_name in enumerate(classes):

    class_path = os.path.join(dataset_path, class_name)

    images = [
        f for f in os.listdir(class_path)
        if f.lower().endswith(valid_extensions)
    ]

    selected_images = random.sample(images, 5)

    for j, image_name in enumerate(selected_images):

        image_path = os.path.join(class_path, image_name)

        image = Image.open(image_path).convert("RGB")

        plt.subplot(2, 5, i * 5 + j + 1)
        plt.imshow(image)
        plt.axis("off")

        if j == 0:
            plt.title(class_name, fontsize=14)

plt.suptitle("Laptop vs Not Laptop Dataset", fontsize=18)
plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import train_test_split

train_data = []
val_data = []
test_data = []

for class_name in classes:

    class_path = os.path.join(dataset_path, class_name)

    images = [
        os.path.join(class_path, f)
        for f in os.listdir(class_path)
        if f.lower().endswith(valid_extensions)
    ]

    train_images, temp_images = train_test_split(
        images,
        test_size=0.30,
        random_state=42
    )

    val_images, test_images = train_test_split(
        temp_images,
        test_size=0.50,
        random_state=42
    )

    train_data.extend([(img, class_name) for img in train_images])
    val_data.extend([(img, class_name) for img in val_images])
    test_data.extend([(img, class_name) for img in test_images])

print("Dataset split completed!")
print("-" * 45)

for name, data in [
    ("Train", train_data),
    ("Validation", val_data),
    ("Test", test_data)
]:
    print(f"{name}: {len(data)} images")

print("-" * 45)

for class_name in classes:
    print(f"\n{class_name}:")
    print("Train:", sum(1 for _, c in train_data if c == class_name))
    print("Validation:", sum(1 for _, c in val_data if c == class_name))
    print("Test:", sum(1 for _, c in test_data if c == class_name))
```

```python
from tensorflow.keras.utils import load_img, img_to_array

IMG_SIZE = (224, 224)

class_names = ["Laptops", "Not_Laptops"]

class_to_label = {
    "Laptops": 0,
    "Not_Laptops": 1
}

def load_dataset(data):
    images = []
    labels = []

    for image_path, class_name in data:

        image = load_img(
            image_path,
            target_size=IMG_SIZE
        )

        image_array = img_to_array(image)

        image_array = image_array / 255.0

        images.append(image_array)
        labels.append(class_to_label[class_name])

    return np.array(images), np.array(labels)


X_train, y_train = load_dataset(train_data)
X_val, y_val = load_dataset(val_data)
X_test, y_test = load_dataset(test_data)

print("Preprocessing completed!")
print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("X_test:", X_test.shape)

print("\nPixel value range:")
print("Min:", X_train.min())
print("Max:", X_train.max())
```

```python
model = keras.Sequential([
    layers.Input(shape=(224, 224, 3)),

    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(128, (3, 3), activation='relu'),
    layers.MaxPooling2D((2, 2)),

    layers.Flatten(),

    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),

    layers.Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
EPOCHS = 15
BATCH_SIZE = 16

history_baseline = model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    verbose=1
)
```

```python
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)

plt.plot(history_baseline.history['accuracy'],
         label='Training Accuracy')

plt.plot(history_baseline.history['val_accuracy'],
         label='Validation Accuracy')

plt.title('Baseline CNN Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)

plt.plot(history_baseline.history['loss'],
         label='Training Loss')

plt.plot(history_baseline.history['val_loss'],
         label='Validation Loss')

plt.title('Baseline CNN Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

test_loss, test_accuracy = model.evaluate(
    X_test, y_test, verbose=0
)

y_prob = model.predict(X_test)
y_pred = (y_prob > 0.5).astype(int).flatten()

print("Test Loss:", test_loss)
print("Test Accuracy:", test_accuracy)

print("\nClassification Report:")

print(classification_report(
    y_test,
    y_pred,
    target_names=["Laptop", "Not Laptop"]
))

cm = confusion_matrix(y_test, y_pred)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Laptop", "Not Laptop"]
)

disp.plot(cmap="Blues")
plt.title("Baseline CNN Confusion Matrix")
plt.show()
```

```python
def make_gradcam_heatmap(img_array, model):

    base_model = model.get_layer("mobilenetv2_1.00_224")

    conv_model = keras.Model(
        inputs=base_model.inputs,
        outputs=base_model.get_layer("out_relu").output
    )

    rescaling = model.get_layer("rescaling")
    preprocessing = model.get_layer("lambda")

    pooling = model.get_layer("global_average_pooling2d")
    dense = model.get_layer("dense_2")
    final_layer = model.get_layer("dense_3")

    with tf.GradientTape() as tape:

        x = rescaling(img_array)
        x = preprocessing(x)

        conv_outputs = conv_model(x, training=False)

        tape.watch(conv_outputs)

        x = pooling(conv_outputs)
        x = dense(x)
        predictions = final_layer(x)

        loss = predictions[:, 0]

    grads = tape.gradient(loss, conv_outputs)

    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))

    conv_outputs = conv_outputs[0]

    heatmap = tf.reduce_sum(conv_outputs * pooled_grads, axis=-1)

    heatmap = tf.maximum(heatmap, 0)
    heatmap = heatmap / (tf.reduce_max(heatmap) + 1e-8)

    return heatmap.numpy()

```

```python
def make_gradcam_heatmap_efficientnet(img_array, model):

    base_model = model.get_layer("efficientnetb0")

    conv_model = keras.Model(
        inputs=base_model.inputs,
        outputs=base_model.get_layer("top_activation").output
    )

    pooling = model.get_layer("global_average_pooling2d_2")
    dense = model.get_layer("dense_6")
    final_layer = model.get_layer("dense_7")

    with tf.GradientTape() as tape:

        # EfficientNet receives pixel values in range 0-255
        x = img_array * 255.0

        conv_outputs = conv_model(x, training=False)
        tape.watch(conv_outputs)

        x = pooling(conv_outputs)
        x = dense(x)
        predictions = final_layer(x)

        loss = predictions[:, 0]

    grads = tape.gradient(loss, conv_outputs)

    pooled_grads = tf.reduce_mean(grads, axis=(0, 1, 2))

    conv_outputs = conv_outputs[0]

    heatmap = tf.reduce_sum(conv_outputs * pooled_grads, axis=-1)

    heatmap = tf.maximum(heatmap, 0)
    heatmap = heatmap / (tf.reduce_max(heatmap) + 1e-8)

    return heatmap.numpy()
```

```python
import numpy as np
from PIL import Image

def predict_image(image):

    # Convert image to RGB
    image = image.convert("RGB")

    # Resize image
    image = image.resize((224, 224))

    # Convert to array
    img_array = np.array(image).astype("float32") / 255.0

    # Add batch dimension
    img_array = np.expand_dims(img_array, axis=0)

    # Make prediction
    prediction = final_model.predict(img_array, verbose=0)[0][0]

    # Class prediction
    if prediction >= 0.5:
        label = "Not Laptop"
        confidence = prediction
    else:
        label = "Laptop"
        confidence = 1 - prediction

    return {
        label: float(confidence)
    }
```

```python
def gradio_predict_with_gradcam(image):

    if image is None:
        return {"Please upload an image": 1.0}, None

    # Prepare image
    image = image.convert("RGB")
    resized_image = image.resize((224, 224))

    img_array = np.array(resized_image).astype("float32") / 255.0
    img_array = np.expand_dims(img_array, axis=0)

    # Prediction
    prediction = final_model.predict(img_array, verbose=0)[0][0]

    if prediction >= 0.5:
        label = "Not Laptop"
        confidence = float(prediction)
    else:
        label = "Laptop"
        confidence = float(1 - prediction)

    result = {label: confidence}

    # Generate Grad-CAM
    heatmap = make_gradcam_heatmap(img_array, final_model)

    heatmap_resized = cv2.resize(heatmap, (224, 224))

    heatmap_colored = cv2.applyColorMap(
        np.uint8(255 * heatmap_resized),
        cv2.COLORMAP_JET
    )

    heatmap_colored = cv2.cvtColor(
        heatmap_colored,
        cv2.COLOR_BGR2RGB
    )

    # Overlay
    original_img = img_array[0]

    overlay = (
        0.6 * original_img +
        0.4 * heatmap_colored / 255.0
    )

    overlay = np.clip(overlay, 0, 1)
    overlay = (overlay * 255).astype("uint8")

    overlay_image = Image.fromarray(overlay)

    return result, overlay_image


demo = gr.Interface(
    fn=gradio_predict_with_gradcam,
    inputs=gr.Image(type="pil", label="Upload an Image"),
    outputs=[
        gr.Label(num_top_classes=2, label="Prediction"),
        gr.Image(type="pil", label="Grad-CAM Explanation")
    ],
    title="Laptop vs Not Laptop Classifier",
    description="Upload an image to classify it and visualize which image regions influenced the prediction."
)

demo.launch(share=True)
```
